```python
import pandas as pd

# Your dataframe `df` already contains the necessary columns
def prepare_data_for_t5(df):
    data = []
    for i in range(len(df)):
        entity_name = df.loc[i, "entity_name"]
        ocr_text = df.loc[i, "ocr_text"]
        entity_value = df.loc[i, "entity_value"]

        # Create the prompt (input)
        input_text = f"Extract the {entity_name} from the item description: {ocr_text} for this product."
        
        # Target is the actual entity value
        target_text = entity_value

        data.append((input_text, target_text))

    return data

# Prepare the data
prepared_data = prepare_data_for_t5(df)
```

```python
from transformers import T5Tokenizer, T5ForConditionalGeneration, AdamW
import torch
from torch.utils.data import Dataset, DataLoader
import sentencepiece

# Custom dataset class for T5 fine-tuning
class T5Dataset(Dataset):
    def __init__(self, data, tokenizer, max_length=512):
        self.data = data
        self.tokenizer = tokenizer
        self.max_length = max_length

    def __len__(self):
        return len(self.data)

    def __getitem__(self, index):
        input_text, target_text = self.data[index]

        # Tokenize input and target texts
        input_ids = self.tokenizer(input_text, return_tensors="pt", padding="max_length", truncation=True, max_length=self.max_length).input_ids.squeeze()
        target_ids = self.tokenizer(target_text, return_tensors="pt", padding="max_length", truncation=True, max_length=self.max_length).input_ids.squeeze()

        return {
            "input_ids": input_ids,
            "target_ids": target_ids
        }

# Initialize T5 model and tokenizer
tokenizer = T5Tokenizer.from_pretrained("google/flan-t5-large")
model = T5ForConditionalGeneration.from_pretrained("google/flan-t5-large")

# Prepare dataset
t5_dataset = T5Dataset(prepared_data, tokenizer)
dataloader = DataLoader(t5_dataset, batch_size=8, shuffle=True)

# Set up optimizer
optimizer = AdamW(model.parameters(), lr=5e-5)

# Move model to GPU if available
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model = model.to(device)

# Training loop
epochs = 3
for epoch in range(epochs):
    model.train()
    total_loss = 0
    for batch in dataloader:
        torch.cuda.empty_cache()
        optimizer.zero_grad()

        # Move input and target tensors to the same device as the model
        input_ids = batch["input_ids"].to(device)
        target_ids = batch["target_ids"].to(device)
        
        # Forward pass
        outputs = model(input_ids=input_ids, labels=target_ids)
        loss = outputs.loss

        # Backpropagation
        loss.backward()
        optimizer.step()

        total_loss += loss.item()

    print(f"Epoch {epoch + 1}/{epochs} - Loss: {total_loss / len(dataloader)}")

# Save the fine-tuned model
model.save_pretrained("fine-tuned-flan-t5")
tokenizer.save_pretrained("fine-tuned-flan-t5")
```

```python
# Load the fine-tuned model
tokenizer = T5Tokenizer.from_pretrained("fine-tuned-flan-t5")
model = T5ForConditionalGeneration.from_pretrained("fine-tuned-flan-t5")

# Example input
input_text = "Extract the item_weight from the item description: COMPOSITIONServing Size: (0.709 g) for this product."

# Tokenize input
input_ids = tokenizer(input_text, return_tensors="pt").input_ids

# Generate output
outputs = model.generate(input_ids)
output_text = tokenizer.decode(outputs[0], skip_special_tokens=True)

print("Extracted Entity Value:", output_text)
```
